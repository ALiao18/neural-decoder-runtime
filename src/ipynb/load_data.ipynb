# Load Data and look at shape

```python
import numpy as np 
import scipy.io
```

```python
path = '/Users/aliao/Documents/neural-decoder-runtime/data/willet/competitionData/train/t12.2022.04.28.mat'
data = scipy.io.loadmat(path)
print(data.keys())
```

```python
spike_pow = data['spikePow']
tx1 = data['tx1']
block_idx = data['blockIdx']
sentences = data['sentenceText']

print(f"Number of sentences: {spike_pow.shape[0]}")
print(f"First trial spikePow shape: {spike_pow[0,0].shape}")
print(f"First trial tx1 shape: {tx1[0,0].shape}")
print(f"Block indices: {np.unique(block_idx)}")
print(f"First sentence: {sentences[0].tobytes().decode('utf-16').strip()}")
```

```python
print(f"Number of sentences: {spike_pow.shape[1]}")
print(f"Trial 0 shape: {spike_pow[0,0].shape}")
print(f"Trial 1 shape: {spike_pow[0,1].shape}")
print(f"Trial 2 shape: {spike_pow[0,2].shape}")
print(f"Trial 3 shape: {spike_pow[0,3].shape}")
```

```python
print(f"blockIdx shape: {block_idx.shape}")
print(f"blockIdx dtype: {block_idx.dtype}")
print(f"First 5 block indices: {block_idx[0,:5]}")
```

```python
print(f"Unique blocks: {np.unique(block_idx)}")
print(f"Sentences per block: {dict(zip(*np.unique(block_idx, return_counts=True)))}")
```
